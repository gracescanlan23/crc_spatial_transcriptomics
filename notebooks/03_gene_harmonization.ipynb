# 03. Gene Harmonization Between Xenium and scRNA-seq

## Objective

This notebook evaluates gene overlap between the QC-filtered
Xenium spatial transcriptomics dataset and the matched
colorectal cancer scRNA-seq reference.

The analysis will:

1. Load both processed datasets.
2. Examine gene identifiers and naming consistency.
3. Identify genes shared between Xenium and scRNA-seq.
4. Characterize shared and dataset-specific genes.
5. Prepare the shared gene set for downstream imputation.
6. Save gene harmonization results.

The shared gene set will provide the foundation for
held-out gene validation during spatial gene imputation.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import scanpy as sc
import matplotlib.pyplot as plt

# Project directories
PROJECT_ROOT = Path.cwd().parent

PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"
TABLE_DIR = PROJECT_ROOT / "results" / "tables"
FIGURE_DIR = PROJECT_ROOT / "results" / "figures"

TABLE_DIR.mkdir(parents=True, exist_ok=True)
FIGURE_DIR.mkdir(parents=True, exist_ok=True)

# Load QC-filtered Xenium dataset
adata_xenium = sc.read_h5ad(
    PROCESSED_DIR / "xenium_colon_qc_filtered.h5ad"
)

# Load matched tumor scRNA-seq reference
adata_scrna = sc.read_h5ad(
    PROCESSED_DIR / "GSE289314_matched_tumor_reference.h5ad"
)

print("XENIUM DATASET")
print(adata_xenium)

print("\nscRNA-seq REFERENCE")
print(adata_scrna)
```
